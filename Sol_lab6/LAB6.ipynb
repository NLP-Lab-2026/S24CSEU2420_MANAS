# Q1. Implement Word2Vec using CBOW



```python
english_corpus = [
    "Machine learning is a branch of artificial intelligence.",
    "Artificial intelligence enables machines to learn.",
    "Deep learning uses neural networks for complex tasks.",
    "Neural networks are inspired by the human brain.",
    "Computers process data using algorithms.",
    "Python is a popular programming language.",
    "Java is widely used for software development.",
    "Programming helps us solve computational problems.",
    "Data science combines statistics and computer science.",
    "Machine learning models learn patterns from data.",
    "Artificial intelligence is transforming modern technology.",
    "Deep learning models require large amounts of data.",
    "Python supports machine learning libraries.",
    "Neural networks contain interconnected artificial neurons.",
    "Algorithms help computers perform specific tasks.",
    "Computer science studies computation and information.",
    "Software developers write code to build applications.",
    "Data analysis helps organizations make decisions.",
    "Machine learning is used in healthcare.",
    "Artificial intelligence is used in robotics.",
    "Robots use sensors to understand their surroundings.",
    "Natural language processing helps computers understand text.",
    "Language models learn from large text collections.",
    "Word embeddings represent words as numerical vectors.",
    "Word2Vec is a technique for learning word embeddings.",
    "CBOW predicts a target word from surrounding words.",
    "Skip Gram predicts surrounding words from a target word.",
    "GloVe learns word representations using global statistics.",
    "Cosine similarity measures similarity between vectors.",
    "Text preprocessing improves natural language processing.",
    "Tokenization divides text into individual words.",
    "Stopwords are common words in natural language.",
    "Stemming reduces words to shorter root-like forms.",
    "Lemmatization converts words into their base forms.",
    "Search engines retrieve relevant documents.",
    "Information retrieval finds useful information from text.",
    "Chatbots communicate with users through natural language.",
    "Artificial intelligence powers intelligent applications.",
    "Data scientists analyze large datasets.",
    "Machine learning can classify images.",
    "Computer vision enables machines to interpret images.",
    "Speech recognition converts spoken language into text.",
    "Deep learning supports speech recognition systems.",
    "Cloud computing provides computing resources online.",
    "Cybersecurity protects computer systems from threats.",
    "Databases store and organize information.",
    "Big data contains large and complex datasets.",
    "Optimization improves the performance of algorithms.",
    "Training data helps models learn useful patterns.",
    "Word embeddings capture relationships between words."
]

assert len(english_corpus) >= 50
def preprocess_english(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", " ", text)
    return text.split()

english_tokens = [preprocess_english(s) for s in english_corpus]
print("Sentence count:", len(english_corpus))
print("First original sentence:", english_corpus[0])
print("First tokenized sentence:", english_tokens[0])
```

```python
# CBOW is selected with sg=0.
cbow_model = Word2Vec(
    sentences=english_tokens,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0,
    workers=1,
    seed=42,
    epochs=100
)

print("CBOW training complete")
print("Vocabulary size:", len(cbow_model.wv))
print("Vector dimensionality:", cbow_model.wv.vector_size)
```

```python

# Display 10 sample word vectors and their dimensions.
sample_words = [
    "machine", "learning", "python", "java", "computer",
    "data", "artificial", "intelligence", "neural", "networks"
]
for word in sample_words:
    print(f"\nWord: {word}")
    print("Vector:", np.round(cbow_model.wv[word], 4))
    print("Dimensions:", cbow_model.wv[word].shape)
```

```python
# Cosine similarity for five word pairs. Use words actually in the vocabulary.
word_pairs_q1 = [
    ("machine", "learning"),
    ("python", "java"),
    ("artificial", "intelligence"),
    ("data", "science"),
    ("neural", "networks")
]
for w1, w2 in word_pairs_q1:
    print(f"{w1} — {w2}: {cbow_model.wv.similarity(w1, w2):.4f}")
```

```python
# Nearest neighbours for five selected words.
for word in ["machine", "python", "computer", "learning", "data"]:
    print(f"\nMost similar to {word}:")
    print(cbow_model.wv.most_similar(word, topn=5))
```

```python
hindi_model = Word2Vec(
    sentences=hindi_tokens,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0,
    workers=1,
    seed=42,
    epochs=100
)
print("Hindi CBOW model trained.")
print("Vocabulary size:", len(hindi_model.wv))
print("Vector dimensions:", hindi_model.wv.vector_size)

# Consistent normalization lookup avoids Unicode key mismatches.
hindi_vocab_normalized = {
    unicodedata.normalize("NFC", word): word
    for word in hindi_model.wv.index_to_key
}
def find_hindi_word(word):
    return hindi_vocab_normalized.get(unicodedata.normalize("NFC", word))

print("मशीन found:", find_hindi_word("मशीन") is not None)
```

```python
# Sentence vector = mean of the embeddings of the words in that sentence.
def hindi_sentence_vector(tokens):
    vectors = [
        hindi_model.wv[find_hindi_word(token)]
        for token in tokens
        if find_hindi_word(token) is not None
    ]
    if not vectors:
        return np.zeros(hindi_model.wv.vector_size, dtype=float)
    return np.mean(vectors, axis=0)

hindi_sentence_vectors = np.vstack([
    hindi_sentence_vector(tokens) for tokens in hindi_tokens
])
print("Number of sentence vectors:", len(hindi_sentence_vectors))
print("Each sentence vector shape:", hindi_sentence_vectors[0].shape)
print("First sentence vector:", np.round(hindi_sentence_vectors[0], 4))
```

# Q4. Parallel corpus and cross-lingual information retrieval


```python
parallel_pairs = [
    ("Good morning.", "सुप्रभात।"),
    ("How are you?", "आप कैसे हैं?"),
    ("My name is Ravi.", "मेरा नाम रवि है।"),
    ("I am a student.", "मैं एक विद्यार्थी हूँ।"),
    ("I study computer science.", "मैं कंप्यूटर विज्ञान पढ़ता हूँ।"),
    ("Machine learning is useful.", "मशीन लर्निंग उपयोगी है।"),
    ("Artificial intelligence is changing the world.", "कृत्रिम बुद्धिमत्ता दुनिया को बदल रही है।"),
    ("The sun gives us light.", "सूर्य हमें प्रकाश देता है।"),
    ("Water is necessary for life.", "पानी जीवन के लिए आवश्यक है।"),
    ("Trees provide shade.", "पेड़ छाया प्रदान करते हैं।"),
    ("India is a diverse country.", "भारत विविधताओं वाला देश है।"),
    ("Delhi is the capital of India.", "दिल्ली भारत की राजधानी है।"),
    ("Books give us knowledge.", "किताबें हमें ज्ञान देती हैं।"),
    ("Exercise keeps the body healthy.", "व्यायाम शरीर को स्वस्थ रखता है।"),
    ("The teacher explains the lesson.", "शिक्षक पाठ समझाते हैं।"),
    ("The farmer grows crops.", "किसान फसल उगाता है।"),
    ("The doctor treats patients.", "डॉक्टर मरीजों का इलाज करता है।"),
    ("Technology makes life easier.", "तकनीक जीवन को आसान बनाती है।"),
    ("The computer processes data.", "कंप्यूटर डेटा संसाधित करता है।"),
    ("Hard work leads to success.", "मेहनत सफलता की ओर ले जाती है।"),
    ("The family supports each other.", "परिवार एक दूसरे का सहयोग करता है。"),
    ("Music brings peace to the mind.", "संगीत मन को शांति देता है।")
]
english_queries = [p[0] for p in parallel_pairs]
hindi_candidates = [p[1] for p in parallel_pairs]
print("Aligned sentence pairs:", len(parallel_pairs))
print("Example:", parallel_pairs[0])
```

```python
from sentence_transformers import SentenceTransformer

MODEL_NAME = "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
try:
    print("Loading multilingual sentence embedding model...")
    multilingual_model = SentenceTransformer(MODEL_NAME)
    print("Multilingual model loaded.")
except Exception as e:
    multilingual_model = None
    print("Could not load multilingual model:", repr(e))
    print("Check internet access and rerun this cell.")
```

```python
if multilingual_model is not None:
    english_embeddings = multilingual_model.encode(
        english_queries, convert_to_numpy=True, normalize_embeddings=True
    )
    hindi_embeddings = multilingual_model.encode(
        hindi_candidates, convert_to_numpy=True, normalize_embeddings=True
    )
    print("English embedding matrix:", english_embeddings.shape)
    print("Hindi embedding matrix:", hindi_embeddings.shape)
else:
    print("Skipped embeddings because the model did not load.")
```

```python
if multilingual_model is not None:
    # Rows are English queries; columns are Hindi candidates.
    similarity_matrix = english_embeddings @ hindi_embeddings.T
    print("Similarity matrix shape:", similarity_matrix.shape)

    # Evaluate at least five queries and show their ranked Hindi candidates.
    number_to_show = 5
    rank1_hits = 0
    rank5_hits = 0

    for query_idx in range(number_to_show):
        scores = similarity_matrix[query_idx]
        ranked_indices = np.argsort(scores)[::-1]
        correct_index = query_idx  # Parallel corpus is aligned by index.
        rank_of_correct = int(np.where(ranked_indices == correct_index)[0][0]) + 1

        if rank_of_correct == 1:
            rank1_hits += 1
        if rank_of_correct <= 5:
            rank5_hits += 1

        print("\n" + "=" * 70)
        print("English query:", english_queries[query_idx])
        print("Correct aligned Hindi sentence:", hindi_candidates[correct_index])
        print("Correct sentence rank:", rank_of_correct)
        print("Correct sentence similarity:", f"{scores[correct_index]:.4f}")
        print("Top 5 retrieved Hindi candidates:")
        for rank, candidate_idx in enumerate(ranked_indices[:5], start=1):
            print(
                f"Rank {rank}: {hindi_candidates[candidate_idx]} "
                f"(score={scores[candidate_idx]:.4f})"
            )

    recall_at_1 = rank1_hits / number_to_show
    recall_at_5 = rank5_hits / number_to_show
    print("\nEvaluation on first five queries")
    print(f"Recall@1: {recall_at_1:.2f} ({rank1_hits}/{number_to_show})")
    print(f"Recall@5: {recall_at_5:.2f} ({rank5_hits}/{number_to_show})")
else:
    print("Retrieval skipped because the multilingual model did not load.")
```
